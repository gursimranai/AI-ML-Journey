# Q5. Evaluate and Print the Result of the Expression

## 🎯 Objective

Evaluate the following Python expression and explain **why the output is what it is** based on **Python's operator precedence**.

---


````python
x = 10 + 3 * 2 ** 2
print(x)
````

# 🔍 Step-by-Step Evaluation

Python evaluates expressions according to **operator precedence** (order of operations).

| Priority | Operator | Description |
|----------|----------|-------------|
| 1️⃣ | `()` | Parentheses |
| 2️⃣ | `**` | Exponentiation (Power) |
| 3️⃣ | `*`, `/`, `//`, `%` | Multiplication, Division, Floor Division, Modulus |
| 4️⃣ | `+`, `-` | Addition, Subtraction |

---

## Step 1: Exponentiation (`**`)

```python
2 ** 2
```

Result:

```python
4
```

The expression becomes:

```python
10 + 3 * 4
```

---

## Step 2: Multiplication (`*`)

```python
3 * 4
```

Result:

```python
12
```

The expression becomes:

```python
10 + 12
```

---

## Step 3: Addition (`+`)

```python
10 + 12
```

Result:

```python
22
```

---

# 📊 Evaluation Table

| Step | Expression | Result |
|------|------------|--------|
| Original | `10 + 3 * 2 ** 2` | |
| 1 | `10 + 3 * 4` | `2 ** 2 = 4` |
| 2 | `10 + 12` | `3 * 4 = 12` |
| 3 | `22` | `10 + 12 = 22` |

---

# 💡 Why is the Output 22?

Python does **not** evaluate expressions from left to right. Instead, it follows **operator precedence**:

- First, calculate the exponent (`2 ** 2 = 4`).
- Next, perform multiplication (`3 × 4 = 12`).
- Finally, perform addition (`10 + 12 = 22`).

Therefore, the final value stored in `x` is:

```python
x = 22
```

---


## 📚 Key Takeaways

- `**` (Exponentiation) has **higher precedence** than `*`.
- `*` (Multiplication) has **higher precedence** than `+`.
- Python always evaluates operators according to its **operator precedence rules**, unless parentheses `()` are used to change the order.

> **Final Result:** **`x = 22`** ✅
