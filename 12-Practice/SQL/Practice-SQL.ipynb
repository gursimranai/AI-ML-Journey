# What is Databases?
> **Databases** is the collection of data in the format that can be easily accessed.

* Why **SQL**?
   - Can store large data.
   - feature like security and scalability.
   - Easier to insert, search, update or delete data.
* DBMS is a layer to SQL , like for Queries etc.

# SQL

- **Relational Databases** => Data Stored in Table format.
- **Examples:-** MySQL, Oracle, PostgreSQL.

# NoSQL

- **Non-Relational-Databases** => Data stored in document, key-value, graphs etc.
- **Examples:-** MongoDB, Cassandra, Neo4j etc.

> **Inside MySQL** - We can Create Multible Databases. <br>
> **Inside Databases** -  We can create Multimple Tables.

# SQL (Structured Query Language)

- SQL is a programming language used to interact with relational databases.
