# Python List Vs Numpy Array

- Numpy Array are faster (Because numpy optimized in c Language)
- Numpy store element in contiguous memory
- use less memory
- build in mathematical function
- support direct vectorized operation

```python
# Execution speed
import time

py_list = list(range(1_000_000))
start = time.time()
sq_list = [x for x in py_list]
end = time.time()

print(f"The time taken by list is : {end-start} seconds")

np_arr = np.arange(1_000_000)
start = time.time()
sq_arr = np_arr ** 2
end = time.time()
print(f"The time taken by numpy array is : {end-start} secconds")
```

```python
# Memory

import sys

print(f"size of list = {sys.getsizeof(py_list) * len(py_list)} bytes")
print(f"size of array = {np_arr.nbytes}")
```
