```python
# Pandas `parse_dates` Markdown Note (Triple Quotes Format)


"""
# Pandas `parse_dates` Parameter 🗓️

## What is `parse_dates`?

The `parse_dates` parameter in `pd.read_csv()` automatically converts
date columns from strings (object) to Pandas datetime format (datetime64[ns]).

--------------------------------------------------

## Syntax

pd.read_csv(
    "file.csv",
    parse_dates=["Date"]
)

--------------------------------------------------

## Example 1: Parse a Single Date Column

CSV:

Date,Sales
2026-08-14,100
2026-08-15,120

Code:

df = pd.read_csv(
    "sales.csv",
    parse_dates=["Date"]
)

print(df.dtypes)

Output:

Date     datetime64[ns]
Sales             int64

✅ The Date column is automatically converted to datetime.

--------------------------------------------------

## Example 2: Parse Multiple Date Columns

df = pd.read_csv(
    "data.csv",
    parse_dates=["Start_Date", "End_Date"]
)

✅ Both columns are converted to datetime format.

--------------------------------------------------

## Example 3: Combine Multiple Columns into One Date

CSV:

Year,Month,Day,Value
2026,8,14,100
2026,8,15,120

Code:

df = pd.read_csv(
    "data.csv",
    parse_dates=[["Year", "Month", "Day"]]
)

Output:

Year_Month_Day    Value
2026-08-14        100
2026-08-15        120

✅ Pandas combines the columns into a single datetime column.

--------------------------------------------------

## Without parse_dates

df = pd.read_csv("data.csv")

Output:

Date    object

Manual Conversion:

df["Date"] = pd.to_datetime(df["Date"])

--------------------------------------------------

## Quick Revision

parse_dates=["Date"]
→ Convert one column to datetime

parse_dates=["A", "B"]
→ Convert multiple columns to datetime

parse_dates=[["Year", "Month", "Day"]]
→ Combine multiple columns into one datetime column

pd.to_datetime()
→ Manual datetime conversion

--------------------------------------------------


"""
```

```python
# Label Encoding:

import pandas as pd
from sklearn.preprocessing import LabelEncoder

# 1. Create a sample dataset
data = {'Customer_ID': [101, 102, 103, 104],
        'Country': ['USA', 'India', 'France', 'USA']}
df = pd.DataFrame(data)

# 2. Initialize the LabelEncoder
encoder = LabelEncoder()

# 3. Fit and transform the categorical column
df['Country_Encoded'] = encoder.fit_transform(df['Country'])

# View results
print(df)
```

```python
# Using the Interquartile Range (IQR):

Q1 = data['column_name'].quantile(0.25)
Q3 = data['column_name'].quantile(0.75)
IQR = Q3 - Q1
data = data[(data['column_name'] >= Q1 - 1.5 * IQR) & (data['column_name'] <= Q3 + 1.5 * IQR)]
```
