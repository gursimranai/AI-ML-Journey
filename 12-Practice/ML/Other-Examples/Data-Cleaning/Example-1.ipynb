# Import Useful Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# Load Dataset

```python
url = "https://people.sc.fsu.edu/~jburkardt/data/csv/hw_200.csv"
df = pd.read_csv(url)
df.columns = ["Index", "Height", "Weight"]
```

```python
df
```

```python
# Data set Summary

print(df.describe())
```

# Creating Visualization


```python
# Scatter plot
plt.figure(figsize=(8,6))
plt.scatter(df["Height"],df["Weight"],color='blue', alpha=0.5,)
plt.title("Height vs Weight",fontsize=16)
plt.xlabel("Height (inches)",fontsize=12)
plt.ylabel("Weight (pounds)",fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

# Handle missing value

```python
# if df.isnull().values.any():
#     df.fillna(df.mean(),inplace =True)
```

```python
# Second Way to check for missing values and fill them with mean
df.isnull().sum()
```

```python
# there are no missing values in the dataset, so we don't need to fill any values with mean.
# if any column had missing values, we could fill them with the mean of that column using the following code:
df['Index'] = df["Index"].fillna(df["Index"].mean())
```

# Adding New Column

```python

df["BMI"] = df["Weight"] / (df["Height"] / 39.37) ** 2
```

```python
print(df.head())
```

```python
# This Regression plot will show the relationship between Height and Weight,
# along with a regression line that best fits the data points. (Study Latter)
# The scatter points will be in blue, and the regression line will be in red.

sns.regplot(
    x="Height",
    y="Weight",
    data=df,
    scatter_kws={"color": "blue"},
    line_kws={"color": "red"}
)

plt.title("Height vs Weight with Regression Line")

# Saving plot tight is used to ensure that the saved image does not have any extra whitespace around it.
plt.savefig(
    "Height_vs_weight_plot_with_regression.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
```
