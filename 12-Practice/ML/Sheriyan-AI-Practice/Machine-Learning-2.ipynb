```python
ch_mean = df.loc[df['Cholesterol'] != 0, 'Cholesterol'].mean()
```

```python
df['Cholesterol'] = df['Cholesterol'].replace(0, ch_mean).round(2)
```

```python
rb_mean = df.loc[df['RestingBP'] != 0, 'RestingBP'].mean()

df['RestingBP'] = df['RestingBP'].replace(0, rb_mean).round(2)
```

# Data Preprocessing and cleaning

```python
df_encode = pd.get_dummies(df, drop_first=True)
df_encode = df_encode.astype(int)
```

```python
df_encode.head()
```

```python
from sklearn.preprocessing import StandardScaler

numeric_columns = ['Age', 'RestingBP', 'Cholesterol', 'MaxHR', 'Oldpeak']
scalar = StandardScaler()
df_encode[numeric_columns] = scalar.fit_transform(df_encode[numeric_columns])

```
