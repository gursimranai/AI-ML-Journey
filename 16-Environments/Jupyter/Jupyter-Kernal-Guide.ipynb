# 🤔 Why Do We Need Separate Kernels?

Suppose you are working on two projects.

### 📂 Project A

- Python 3.12
- TensorFlow
- NumPy 1.26

### 📂 Project B

- Python 3.11
- PyTorch
- NumPy 2.x

If both projects share the same Python installation, package versions may conflict.

A **Virtual Environment** solves this problem by creating an isolated Python installation for each project.

Jupyter can connect to these environments using **Kernels**.

# 🧠 What is a Jupyter Kernel?

A **Kernel** is the Python interpreter that executes every code cell inside a notebook.

```

Notebook
      │
      ▼
Jupyter Kernel
      │
      ▼
Python Interpreter
      │
      ▼
Installed Packages

```

Changing the kernel changes the Python interpreter used by the notebook.

````python
# Create a Virtual Environment

python -m venv myenv
````

# ▶ Activate the Environment

````python
# Windows CMD

myenv\Scripts\activate
````

````python
# Windows PowerShell

myenv\Scripts\Activate.ps1
````

````python
# Linux / macOS

source myenv/bin/activate
````

After activation you should see

```text
(myenv)
C:\Projects>
```

This indicates that the virtual environment is active.

# 📦 Install Jupyter Packages

````python
pip install notebook
````

````python
pip install jupyterlab
````

````python
pip install ipykernel
````

# ⭐ Register the Environment as a Jupyter Kernel

Use the following command to make the environment available inside Jupyter.

````python
python -m ipykernel install --user --name=myenv --display-name "Python (MyEnv)"
````

## 🔍 Command Explanation

| Option | Description |
|---------|-------------|
| `--user` | Install only for current user |
| `--name` | Internal kernel identifier |
| `--display-name` | Name shown inside Jupyter |

Example

```

Python (AI)
Python (ML)
Python (TensorFlow)
Python (Deep Learning)

```

# 🚀 Start Jupyter Notebook

````python
jupyter notebook
````

````python
jupyter lab
````

After opening a notebook

```

Kernel
   ↓

   ↓

```

Your notebook is now connected to your virtual environment.

# ✅ Verify the Active Python Environment

````python
import sys

print(sys.executable)
````

Expected Output

```text
C:\Projects\MyProject\myenv\Scripts\python.exe
```

If you see the environment path, the correct kernel is active.


# 🗑 Remove a Kernel

````python
jupyter kernelspec uninstall myenv
````

# ✏ Rename a Kernel

Simply register the environment again with a different display name.

````python
python -m ipykernel install --user \
--name=myenv \
--display-name "Python (Machine Learning)"
````

# 💡 Best Practices

✅ Create one virtual environment per project.

✅ Give kernels meaningful names.

✅ Never install packages globally unless necessary.

✅ Always verify the active interpreter using `sys.executable`.

✅ Save dependencies using `requirements.txt`.

✅ Remove unused kernels to keep Jupyter clean.

````python
# Save all installed packages

pip freeze > requirements.txt
````

````python
# Reinstall packages later

pip install -r requirements.txt
````
