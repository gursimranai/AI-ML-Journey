## 1. Understanding RGB Colour Representation

RGB stands for Red, Green, and Blue.

It is an additive colour model in which different colours
are produced by combining three primary light components.

Each component has an integer value ranging from 0 to 255.

### RGB Components

- Red: Controls the intensity of red.
- Green: Controls the intensity of green.
- Blue: Controls the intensity of blue.

### Examples

| Colour | Red | Green | Blue |
|--------|-----|-------|------|
| Red | 255 | 0 | 0 |
| Green | 0 | 255 | 0 |
| Blue | 0 | 0 | 255 |
| White | 255 | 255 | 255 |
| Black | 0 | 0 | 0 |
| Yellow | 255 | 255 | 0 |
| Cyan | 0 | 255 | 255 |
| Magenta | 255 | 0 | 255 |

For example:

RGB(255, 100, 50)

- Red = 255
- Green = 100
- Blue = 50

The combination of these three values determines the
resulting colour.

## 4. Understanding the sty Library

sty is a third-party Python library used to apply
colours and text formatting in terminal output.

Unlike random, sty is not included in Python's
standard library.

### Installation

Install it using pip:

pip install sty

### Important Components

| Component | Purpose |
|-----------|---------|
| fg | Changes foreground text colour |
| bg | Changes background colour |
| ef | Applies text effects |
| rs | Resets formatting |

### Understanding fg()

The fg() function accepts RGB values and generates
the corresponding terminal foreground colour sequence.

Syntax:

fg(red, green, blue)

Each RGB component should be between 0 and 255.

### Understanding rs.all

Terminal formatting can remain active after printing.
rs.all resets the formatting to its default state.

### Additional Text Effects Using ef

The `ef` component in the `sty` library provides additional
text formatting effects that can change how terminal text
is displayed.

Some commonly used effects include:

| Attribute | Description |
|---|---|
| `ef.bold` | Makes text bold |
| `ef.italic` | Displays italic text |
| `ef.underl` | Underlines text |
| `ef.strike` | Applies strikethrough formatting |
| `ef.blink` | Attempts to display blinking text |
| `ef.inverse` | Swaps foreground and background colours |

### Important Note

Not every terminal supports all text effects.

For example, blinking text may not work in modern
terminal emulators. The actual appearance depends
on terminal capabilities and configuration.

## 6. Generating Random RGB Values

Our first major task is to generate three random
integer values representing Red, Green, and Blue.

We use:

random.randint(0, 255)

### Why randint(0, 255)?

The randint() function generates an integer between
the specified lower and upper boundaries, including
both boundaries.

Since valid RGB components range from 0 to 255,
we use these exact boundaries.

### Implementation

We generate three separate values:

- red
- green
- blue

These values will later be combined to create a colour.

````python
red = random.randint(0, 255)
green = random.randint(0, 255)
blue = random.randint(0, 255)

print("Red:", red)
print("Green:", green)
print("Blue:", blue)
````

## 7. Understanding User-Defined Functions

A function is a reusable block of code designed
to perform a particular task.

Python functions are defined using the def keyword.

### Basic Syntax

def function_name(parameters):
    # Function body
    return result

### Important Function Components

1. def: Defines a function.
2. Function name: Identifies the function.
3. Parameters: Receive input values.
4. Function body: Contains the instructions.
5. return: Sends a result back to the caller.

### Functions in Our Project

We will create two main functions:

1. generate_rgb()
   - Generates three random RGB values.
   - Returns them together.

2. generate_color(red, green, blue)
   - Accepts RGB values.
   - Uses sty.fg() to generate terminal colour formatting.
   - Returns the colour sequence.

````python
def generate_rgb():
    """
    Generate random RGB colour values.

    Returns:
        tuple: Three random integer values representing
               Red, Green, and Blue.
    """

    red = random.randint(0, 255)
    green = random.randint(0, 255)
    blue = random.randint(0, 255)

    return red, green, blue
````

````python
rgb = generate_rgb()

print("Generated RGB:", rgb)
print("Data Type:", type(rgb))
````

## 8. Understanding Tuples and Tuple Unpacking

A tuple is an ordered and immutable collection
of Python elements.

Tuples are created using parentheses, although
parentheses are not always required.

Example:

rgb = (255, 100, 50)

### Returning Multiple Values

When we write:

return red, green, blue

Python packs these values into a tuple.

### Tuple Unpacking

We can assign the tuple elements to individual
variables in one statement.

Example:

red, green, blue = rgb

This is called tuple unpacking.

In our project, tuple unpacking allows us to receive
the three generated RGB components separately.

````python
rgb = generate_rgb()

red, green, blue = rgb

print("Red:", red)
print("Green:", green)
print("Blue:", blue)
````

## 9. Generating Terminal Colours

Now that we can generate random RGB values,
we need to convert those values into terminal
foreground colour formatting.

We use the fg() function from sty.

### Function Parameters

generate_color(red, green, blue)

The function accepts three arguments:

- red: Red component.
- green: Green component.
- blue: Blue component.

These values are passed to fg().

The function returns the generated terminal
colour escape sequence.

````python
def generate_color(red, green, blue):
    """
    Generate a terminal foreground colour using RGB values.

    Args:
        red (int): Red component (0-255).
        green (int): Green component (0-255).
        blue (int): Blue component (0-255).

    Returns:
        str: Terminal colour escape sequence.
    """

    return fg(red, green, blue)
````

````python
color = generate_color(255, 0, 0)

print(color + "This is a red message!" + rs.all)
````

## 10. Understanding the Main Function

The main() function acts as the central entry point
for organizing our application's execution.

Instead of writing all program instructions directly
at the top level, we place the main operations inside
a function.

### Responsibilities

1. Call generate_rgb().
2. Unpack the generated RGB tuple.
3. Call generate_color().
4. Display the coloured message.
5. Display the generated RGB values.

This makes the application easier to read,
maintain, and extend.

````python
def main():

    # Generate random RGB values
    red, green, blue = generate_rgb()

    # Generate terminal colour
    color = generate_color(red, green, blue)

    # Display colourful message
    print(
        color + "I am randomly changing color muahhahahahahahahaha"
        + rs.all
    )

    # Display RGB information
    print(f"\nGenerated RGB Color: ({red}, {green}, {blue})")
````

## 11. Complete Random Color Generator

We have now explored all the individual components
required to build our application.

Let's combine everything into one complete program.

The final implementation includes:

- Required module imports
- RGB generation function
- Terminal colour generation function
- Main application function
- Python main execution guard

````python
import random
from sty import fg, rs


def generate_rgb():
    """
    Generate random RGB colour values.

    Returns:
        tuple: Three random integer values representing
               Red, Green, and Blue.
    """
    red = random.randint(0, 255)
    green = random.randint(0, 255)
    blue = random.randint(0, 255)

    return red, green, blue


def generate_color(red, green, blue):
    """
    Generate a terminal foreground colour using RGB values.

    Args:
        red (int): Red component (0-255).
        green (int): Green component (0-255).
        blue (int): Blue component (0-255).

    Returns:
        str: Terminal colour escape sequence.
    """
    return fg(red, green, blue)


def main():
    """
    Main function to generate and display a random colour.
    """

    # Generate random RGB values
    red, green, blue = generate_rgb()

    # Generate terminal colour
    color = generate_color(red, green, blue)

    # Display colourful message
    print(
        color + "I am randomly changing color muahhahahahahahahaha"
        + rs.all
    )

    # Display RGB information
    print(f"\nGenerated RGB Color: ({red}, {green}, {blue})")


if __name__ == "__main__":
    main()
````

````python
# Experiment 1: Multiple Random Colours

import random
from sty import fg

# 1. Function Definition
def generate_rgb():
    """Generates a tuple of 3 random integers representing RGB values (0-255)."""
    # Random number generation for each color channel
    r = random.randint(0, 255)
    g = random.randint(0, 255)
    b = random.randint(0, 255)
    return r, g, b

def main():
    print("--- Displaying 5 Random Colors ---\n")
    
    # 2. For Loop (Runs exactly 5 times)
    # range(1, 6) will produce numbers 1, 2, 3, 4, 5
    for i in range(1, 6):
        # Call the function to get random RGB values
        r, g, b = generate_rgb()
        
        # Create the sty terminal escape sequence
        color_escape = fg(r, g, b)
        
        # Display the output with individual index tracking
        print(f"Color #{i}: " + color_escape + f"RGB({r}, {g}, {b})" + fg.rs)

if __name__ == "__main__":
    main()
````

````python
# Experiment 2: Random Background Colours

import random
from sty import fg, bg, rs

def generate_rgb():
    """Generates a tuple of 3 random integers representing RGB values (0-255)."""
    r = random.randint(0, 255)
    g = random.randint(0, 255)
    b = random.randint(0, 255)
    return r, g, b

def main():
    print("--- Random Foreground and Background Colors ---\n")
    
    for i in range(1, 6):
        # 1. Generate unique random RGB values for both layers
        fg_r, fg_g, fg_b = generate_rgb()
        bg_r, bg_g, bg_b = generate_rgb()
        
        # 2. Third-party library functions: Convert RGB tuples into ANSI escape sequences
        text_color = fg(fg_r, fg_g, fg_b)
        background_color = bg(bg_r, bg_g, bg_b)
        
        # 3. Terminal formatting: Combine bg and fg, and wipe clean with rs.all
        print(f"Combination #{i}:")
        print(f" -> Text: RGB({fg_r}, {fg_g}, {fg_b})")
        print(f" -> Background: RGB({bg_r}, {bg_g}, {bg_b})")
        print(background_color + text_color + "   Beautiful Styled Terminal Text Output!   " + rs.all)
        print()  # Spacer line

if __name__ == "__main__":
    main()
````

````python
# Experiment 3 : Generate HEX colour code

def generate_hex(red, green, blue):
    """
    Convert RGB values into a hexadecimal colour code.
    """
    return f"#{red:02X}{green:02X}{blue:02X}"


red, green, blue = generate_rgb()

hex_color = generate_hex(red, green, blue)

print("RGB:", (red, green, blue))
print("HEX:", hex_color)
````

````python
# Experiment 4: Continuous Colour Generator

import random
from sty import fg, bg, rs

def generate_rgb():
    """Generates a tuple of 3 random integers representing RGB values (0-255)."""
    r = random.randint(0, 255)
    g = random.randint(0, 255)
    b = random.randint(0, 255)
    return r, g, b

def main():
    print("--- Continuous Color Generator ---")
    print("Press Enter to keep generating new colors, or type 'exit' to quit.\n")
    
    # 1. While Loop for infinite/continuous execution
    while True:
        # Generate new foreground and background variables
        f_r, f_g, f_b = generate_rgb()
        b_r, b_g, b_b = generate_rgb()
        
        # Instantiate escape contexts
        text_color = fg(f_r, f_g, f_b)
        bg_color = bg(b_r, b_g, b_b)
        
        # Display sample output block
        print(bg_color + text_color + f" FG: ({f_r},{f_g},{f_b}) | BG: ({b_r},{b_g},{b_b}) " + rs.all)
        
        # 2. User Input collection point
        user_action = input("Press [Enter] for next color, or type 'exit': ").strip()
        
        # 3. Conditional Statements to check if the exit sentinel is matched
        if user_action.lower() == "exit":
            print("\nExiting application state. Goodbye!")
            break  # Terminates the while loop context instantly

if __name__ == "__main__":
    main()
````

````python
# Experiment 5: User-Controlled Generator


import random
from sty import fg, rs

def generate_rgb():
    """Generates a tuple of 3 random integers representing RGB values (0-255)."""
    r = random.randint(0, 255)
    g = random.randint(0, 255)
    b = random.randint(0, 255)
    return r, g, b

def main():
    print("--- User-Controlled Color Generator ---")
    
    # 1. Take user input
    user_input = input("How many random colors would you like to generate? ").strip()
    
    # 2. Input Validation
    # .isdigit() checks if the string contains only numbers (covers negative numbers and text)
    if user_input.isdigit():
        # 3. Type Conversion (String to Integer)
        count = int(user_input)
        
        if count == 0:
            print("Zero colors requested. Nothing to display!")
            return
            
        print(f"\nGenerating {count} colors:\n")
        
        # 4. Loop execution based on user selection
        for i in range(1, count + 1):
            r, g, b = generate_rgb()
            color_escape = fg(r, g, b)
            
            # Print the colored output
            print(f"Color #{i}: " + color_escape + f"RGB({r}, {g}, {b})" + fg.rs)
            
    else:
        # Error handling for invalid inputs like letters, floats, or negative numbers
        print(fg.red + "Error: Please enter a valid positive whole number!" + fg.rs)

if __name__ == "__main__":
    main()

````
