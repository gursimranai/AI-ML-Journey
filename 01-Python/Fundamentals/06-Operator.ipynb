# 📖 What is an Operator?

An **Operator** is a special symbol that tells Python **what operation should be performed on one or more values.**

Simply,

> **An operator performs an action on data.**

---

# 🧠 Definition

> **Operators are special symbols that perform mathematical, logical, comparison, assignment, or other operations on values called operands.**

---

# 🌍 Real-Life Analogy

Imagine a calculator.

```
10   +   20
│    │    │
│    │    │
│ Operator│
│         │
Operands
```

The calculator understands

```
+
```

means

```
Add both numbers
```

Similarly,

Python also understands operators.

---

# 🧠 What are Operands?

The values on which an operator works are called **Operands**.

Example

```

```

```
+   → Operator
```

When operands and operators combine,

This is an expression.

Another example

```python

```

Another


**Comparison**

# 📝 Assignment Operators
> **Store, Update, and Modify Values**

---

# 📖 What are Assignment Operators?

Assignment Operators are used to **assign (store) values into variables**.

They tell Python:

> **"Take this value and store it inside this variable."**

---

# 🧠 Definition

> **Assignment Operators are used to assign or update values stored in variables.**

---

# 🌍 Real-Life Analogy

Imagine you have an empty box.

```
Empty Box

┌──────────┐
│          │
└──────────┘
```

You put **10 books** inside.

```
books = 10
```

Now,

```
┌──────────┐
│    10    │
└──────────┘
```

The variable **books** now stores the value **10**.

Assignment operators work exactly like putting something into a labeled box.

---

# 📚 Types of Assignment Operators

| Operator | Meaning | Example |
|----------|---------|----------|
| `=` | Assign | `x = 10` |
| `+=` | Add and Assign | `x += 5` |
| `-=` | Subtract and Assign | `x -= 5` |
| `*=` | Multiply and Assign | `x *= 2` |
| `/=` | Divide and Assign | `x /= 2` |
| `//=` | Floor Divide and Assign | `x //= 2` |
| `%=` | Modulus and Assign | `x %= 3` |
| `**=` | Exponent and Assign | `x **= 2` |
| `&=` | Bitwise AND and Assign | `x &= y` |
| `|=` | Bitwise OR and Assign | `x |= y` |
| `^=` | Bitwise XOR and Assign | `x ^= y` |
| `<<=` | Left Shift and Assign | `x <<= 2` |
| `>>=` | Right Shift and Assign | `x >>= 2` |

> 💡 The bitwise assignment operators will be explained in detail in the **Bitwise Operators** section.

---

# 1️⃣ Assignment Operator (`=`)

The **Assignment Operator** stores a value inside a variable.

Syntax

```text
variable = value
```

The value on the **right side** is evaluated first, then stored in the variable on the **left side**.

---

````python
x = 10

print(x)
````

# 2️⃣ Add and Assign (`+=`)

This operator adds a value to the existing variable.

Instead of writing

```python
x = x + 5
```

we simply write

```python
x += 5
```

Both statements produce the same result.

---

````python
x = 10

x += 5

print(x)
````

# 3️⃣ Subtract and Assign (`-=`)

Subtracts a value and stores the result.

---

````python
balance = 500

balance -= 120

print(balance)
````

# 4️⃣ Multiply and Assign (`*=`)

Multiplies the variable by another value.

---

````python
salary = 2500

salary *= 2

print(salary)
````

# 5️⃣ Divide and Assign (`/=`)

Divides the variable and stores the result.

The result is always a **float**.

---

````python
marks = 100

marks /= 4

print(marks)
````

### ⚠️ Important

Even if the answer is a whole number,

`/=` returns a **float**.

---

# 6️⃣ Floor Divide and Assign (`//=`)

Performs floor division and stores the result.

---

````python
number = 17

number //= 5

print(number)
````

# 7️⃣ Modulus and Assign (`%=`)

Stores the remainder after division.

---

# 8️⃣ Exponent and Assign (`**=`)

Raises the variable to a power and stores the result.

---

# 🧠 Compound Assignment Operators

Operators like

```
+=
-=
*=
/=
```

are called **Compound Assignment Operators**.

They perform two operations in one step.

Example

Instead of writing

```python
score = score + 10
```

we write

```python
score += 10
```


# 🧠 Logical Operators
> **Combine Multiple Conditions and Make Smart Decisions**

---

# 📖 What are Logical Operators?

Logical Operators are used to **combine two or more conditions**.

Instead of performing mathematical calculations,

they work with **Boolean values**.

```
True

False
```

---

# 🧠 Definition

> **Logical Operators combine one or more Boolean expressions and return either `True` or `False`.**

---

# 🌍 Real-Life Analogy

Imagine entering a secure building.

To enter, you must have

```
Employee ID

AND

Fingerprint
```

If both are correct

```
Door Opens ✅
```

Otherwise

```
Door Remains Locked ❌
```

Python makes decisions exactly like this.

---

# 📚 Types of Logical Operators

| Operator | Meaning |
|----------|---------|
| `and` | Returns `True` if **both** conditions are True |
| `or` | Returns `True` if **at least one** condition is True |
| `not` | Reverses the Boolean value |

---

# 1️⃣ Logical AND (`and`)

The `and` operator returns **True only when every condition is True**.

If **even one condition is False**,

the result becomes

```
False
```

---

## Truth Table

| Condition A | Condition B | Result |
|-------------|-------------|--------|
| True | True | ✅ True |
| True | False | ❌ False |
| False | True | ❌ False |
| False | False | ❌ False |

---

````python
print(True and True)
````

````python
age = 20
citizen = True

print(age >= 18 and citizen)
````

### 🌍 Real-Life Example

To vote,

a person must

```
Age ≥ 18

AND

Be a Citizen
```

Flow

```
Age ≥18 ?

        │
        ▼

Citizen ?

        │
        ▼

Both True ?

        │
        ▼

Eligible ✅
```

---

# 2️⃣ Logical OR (`or`)

The `or` operator returns **True if at least one condition is True**.

Only when **all conditions are False**

does it return

```
False
```

---

## Truth Table

| Condition A | Condition B | Result |
|-------------|-------------|--------|
| True | True | ✅ True |
| True | False | ✅ True |
| False | True | ✅ True |
| False | False | ❌ False |

---

````python
print(False or False)
````

````python
day = "Sunday"

print(day == "Saturday" or day == "Sunday")
````

### 🌍 Real-Life Example

You can enter a concert if you have

```
VIP Pass

OR

Regular Ticket
```

Only one is enough.

```
VIP ?

      │

OR

      │

Ticket ?

      │
      ▼

Entry Allowed ✅
```

---

# 3️⃣ Logical NOT (`not`)

The `not` operator reverses the Boolean value.

```
True

↓

False
```

```
False

↓

True
```

---

## Truth Table

| Value | Result |
|--------|--------|
| True | False |
| False | True |

---

````python
print(not True)
````

````python
logged_in = False

print(not logged_in)
````
