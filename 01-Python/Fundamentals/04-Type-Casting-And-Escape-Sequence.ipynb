# 🔤 Escape Sequences in Python

---

# 📖 What is an Escape Sequence?

An **Escape Sequence** is a special combination of characters that starts with a **backslash (`\`)**.

It tells Python:

> **"Don't treat the next character as normal text. Give it a special meaning."**

---

# 🌍 Real-Life Analogy

Imagine you're reading a book.

Normally,

```
Hello World
```

is printed as it is.

But suppose you want to start a new line.

You cannot press the **Enter** key inside a string.

Instead, you tell Python:

```
\n
```

Python understands:

> "Move to the next line."

---

# 🔄 How Escape Sequences Work

```
Backslash (\)
      │
      ▼
Special Character
      │
      ▼
Special Action
```

---

# 🌟 1. New Line (`\n`)

Moves the cursor to the next line.

```python
print("Hello\nWorld")
```

Output

```text
Hello
World
```

---

# 🌟 2. Tab (`\t`)

Adds horizontal space (tab).

```python
print("Name\tAge")
```

Output

```text
Name    Age
```

---

# 🌟 3. Backslash (`\\`)

Prints a single backslash.

```python
print("C:\\Users\\John")
```

Output

```text
C:\Users\John
```

---

# 🌟 4. Single Quote (`\'`)

Prints a single quote inside a single-quoted string.

```python
print('It\'s Python')
```

Output

```text
It's Python
```

---

# 🌟 5. Double Quote (`\"`)

Prints double quotes.

```python
print("He said \"Hello\"")
```

Output

```text
He said "Hello"
```

---

# 🌟 6. Backspace (`\b`)

Removes the previous character while printing.

```python
print("ABC\bD")
```

Possible Output

```text
ABD
```

---

# 🌟 7. Carriage Return (`\r`)

Moves the cursor back to the beginning of the same line.

```python
print("Hello\rHi")
```

Possible Output

```text
Hillo
```

(The exact appearance can vary depending on the terminal.)

---

# 🌟 8. Unicode Character (`\u`)

Prints a Unicode character.

```python
print("\u2764")
```

Output

```text
❤
```

---

# 📊 Common Escape Sequences

| Escape Sequence | Meaning |
|-----------------|---------|
| `\n` | New Line |
| `\t` | Tab Space |
| `\\` | Backslash |
| `\'` | Single Quote |
| `\"` | Double Quote |
| `\b` | Backspace |
| `\r` | Carriage Return |
| `\uXXXX` | Unicode Character |

---

# 💡 Raw Strings

Sometimes we don't want Python to process escape sequences.

Use a **raw string** by adding `r` before the string.

```python
print(r"C:\Users\John")
```

Output

```text
C:\Users\John
```

No escaping is needed.

---

# 🎯 Key Takeaways

- Escape sequences begin with `\`.
- They give special meaning to characters.
- They are commonly used for formatting text, file paths, and special symbols.
- Use raw strings (`r"..."`) when you want backslashes to be treated as normal characters.

# 🔄 Type Casting in Python
> *Converting One Data Type into Another*

---

# 📖 What is Type Casting?

**Type Casting** means **converting one data type into another data type**.

Example

```
"100" (String)

↓

100 (Integer)
```

---

# 🌍 Real-Life Analogy

Imagine water.

```
Water
   │
   ├── Ice
   ├── Steam
   └── Liquid
```

The substance is the same,

only its form changes.

Similarly,

```
100

↓

"100"

↓

100.0
```

The value is similar,

but its data type changes.

---

# 🤔 Why Do We Need Type Casting?

Suppose the user enters:

```python
age = input("Enter Age: ")
```

If the user types:

```
20
```

Python stores it as

```python
"20"
```

which is a **string**, not an integer.

To perform calculations,

we convert it.

```python
age = int(age)
```

Now

```
20
```

is an integer.

---

# 🌟 Types of Type Casting

```
Type Casting
      │
 ┌────┴────┐
 │         │
Implicit  Explicit
```

---

# 🌟 1. Implicit Type Casting

Python automatically converts compatible types.

```python
a = 10
b = 2.5

result = a + b

print(result)
```

Output

```text
12.5
```

Python automatically converts `10` into `10.0`.

---

# 🌟 2. Explicit Type Casting

The programmer manually converts the type.

```python
num = "100"

num = int(num)

print(num)
```

Output

```text
100
```

---

# 📊 Common Type Conversion Functions

| Function | Converts To |
|----------|-------------|
| `int()` | Integer |
| `float()` | Float |
| `str()` | String |
| `bool()` | Boolean |
| `list()` | List |
| `tuple()` | Tuple |
| `set()` | Set |
| `dict()` | Dictionary (from key-value pairs) |

---

# 🧩 Examples

### String → Integer

```python
age = "21"

print(int(age))
```

---

### Integer → Float

```python
print(float(10))
```

Output

```text
10.0
```

---

### Float → Integer

```python
print(int(9.8))
```

Output

```text
9
```

> `int()` removes the decimal part; it does **not** round the number.

---

### Integer → String

```python
marks = 95

print(str(marks))
```

Output

```text
"95"
```

---

### String → Boolean

```python
print(bool(""))
print(bool("Python"))
```

Output

```text
False
True
```

An empty string is `False`, while a non-empty string is `True`.

---

# ⚠️ Invalid Type Casting

```python
int("Hello")
```

Output

```text
ValueError
```

Because `"Hello"` is not a valid integer.

---

# 📊 Implicit vs Explicit

| Implicit | Explicit |
|----------|----------|
| Done automatically | Done by programmer |
| No function needed | Uses conversion functions |
| Safer for compatible types | Gives full control |

---

# 🎯 Key Takeaways

- 🔄 Type Casting means converting one data type into another.
- 🤖 **Implicit** casting is performed automatically by Python.
- 👨‍💻 **Explicit** casting is performed manually using functions like `int()`, `float()`, and `str()`.
- ⚠️ Not every conversion is valid. Invalid conversions raise exceptions.
- 🧠 Always check the data type before performing operations.

> **💡 Golden Rule:**  
> **Choose the correct data type before performing calculations. When needed, convert data explicitly using Python's built-in conversion functions.**
