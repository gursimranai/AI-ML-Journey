# 🖨️ Python Print Function - Complete Notes

The `print()` function is used to display output on the screen.

## Basic Syntax

```python
print(object, sep=' ', end='\n', file=sys.stdout, flush=False)
```

### Parameters

| Parameter | Description | Default |
|------------|------------|---------|
| object | Value(s) to print | Required |
| sep | Separator between multiple objects | `' '` |
| end | What to print at the end | `'\n'` |
| file | Output destination | Screen |
| flush | Force output immediately | `False` |

# 🔍 Internal Working of Python print()

The `print()` function is actually a built-in function.

When you write:

```python
print("Hello")
```

Python internally performs:

1. Takes the object(s).
2. Converts them to strings using `str()`.
3. Joins them using the `sep` parameter.
4. Adds the `end` parameter.
5. Sends the final string to `sys.stdout`.
6. Displays it on the terminal/Jupyter output.

---

## Internal Flow

```text
Input Objects
      │
      ▼
Convert using str()
      │
      ▼
Apply sep
      │
      ▼
Apply end
      │
      ▼
Write to sys.stdout
      │
      ▼
Display Output
```

---


## Understanding *objects

```python
print("Python", "AI", "ML")
```

Internally:

```python
objects = ("Python", "AI", "ML")
```

The `*objects` collects all values into a tuple.

---

## Step 1: Convert Objects to String

```python
print(10)
```

Internally:

```python
str(10)
```

Result:

```python
"10"
```

Example:

```python
print(10, 20, True)
```

Internally:

```python
str(10)
str(20)
str(True)
```

Result:

```python
"10"
"20"
"True"
```

---

## Step 2: Apply Separator

```python
print("A", "B", "C", sep="-")
```

Internally:

```python
"A-B-C"
```

Without custom separator:

```python
"A B C"
```

---

## Step 3: Apply End

```python
print("Hello", end="!")
```

Internally:

```python
"Hello!"
```

Default:

```python
"Hello\n"
```

---

## Step 4: Write to stdout

Python sends output to:

```python
sys.stdout
```

Example:

```python
import sys

sys.stdout.write("Hello")
```

Output:

```text
Hello
```

Actually `print()` uses stdout behind the scenes.

---

## print() vs sys.stdout.write()

### print()

```python
print("Hello")
```

Automatically:

- Converts data to string
- Adds newline
- Supports sep
- Supports end

### sys.stdout.write()

```python
import sys

sys.stdout.write("Hello")
```

Requires:

- String only
- No automatic newline

---

## Real Example

```python
print(10, 20, sep="-", end="!")
```

Internal Steps:

1. Convert

```python
"10"
"20"
```

2. Join

```python
"10-20"
```

3. Add end

```python
"10-20!"
```

4. Write

```python
sys.stdout.write("10-20!")
```

Output:

```text
10-20!
```

# Advanced: file Parameter

Default:


Internally:

```python
print("Hello", file=sys.stdout)
```

Print to file:

```python
with open("output.txt", "w") as f:
    print("Hello", file=f)
```

Now output goes to file instead of screen.

# Advanced: flush Parameter

```python
print("Loading...", flush=True)
```

Normally output is buffered.

`flush=True` forces immediate display.

Useful in:

- Progress bars
- Real-time logs
- Streaming output

# Input Statement in Python

## Input Function

- In Python, we take user input using the `input()` function.
- The `input()` function always returns data as a string.
- Therefore, the returned value is usually stored in a variable.

### Syntax

```python
variable = input()
```

### Example

```python
name = input("Enter your name: ")
```

### Output

```text
Enter your name: Gursimran
```

The value entered by the user is stored in the variable `name`.

---

### Input with Message

We can display a message while taking input.

```python
city = input("Enter your city: ")
```

Output:

```text
Enter your city: Patiala
```

---

### Taking Integer Input

Since `input()` returns a string, we need type conversion to store numbers.

```python
age = int(input("Enter your age: "))
```

### Example

```python
age = int(input("Enter age: "))
print(age)
```

---

### Taking Float Input

```python
height = float(input("Enter your height: "))
```

---

### Important Note

```python
x = input()
```

Type of x:

```python
str
```

Even if the user enters:

```text
25
```

Python stores it as:

```python
"25"
```

---

## Summary

| Function | Purpose |
|-----------|----------|
| input() | Takes input from user |
| int() | Converts to integer |
| float() | Converts to float |
| str() | Converts to string |
