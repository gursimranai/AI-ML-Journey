# 📖 What are Comments?

A **comment** is a piece of text inside a Python program that is **ignored by the Python Interpreter** during execution.

Comments are written **for humans**, not for the computer.

They help explain:

- 💡 Why the code exists
- 💡 What a complex block of code is doing
- 💡 Important warnings
- 💡 Future improvements
- 💡 Business logic

> **Definition**
>
> **Comments are non-executable lines used to improve the readability and maintainability of a program.**

---

# 🧠 How Python Sees Comments

```
Python Program
      │
      ▼
Python Interpreter
      │
 ┌────┴───────────┐
 │                │
 ▼                ▼
Code          Comments
 │                │
Execute      Ignore
 │
 ▼
Output
```

Python completely ignores comments.

They never affect program execution.

# 📝 Types of Comments in Python

Python mainly uses four kinds of comments.

```
                Comments
                    │
     ┌──────────────┼──────────────┐
     │              │              │
Single-Line     Multi-Line     Docstrings
     │
Inline Comments
```

---

# 1️⃣ Single-Line Comment

A single-line comment starts with the **hash symbol (`#`)**.

Everything after `#` is ignored by Python.

---

````python
# Print a welcome message
print("Welcome to Python")
````

### Another Example

````python
# Store student's age
age = 20

print(age)
````

# 2️⃣ Inline Comment

Inline comments appear on the **same line** as the code.

They provide a short explanation.

---

````python
age = 18      # Minimum voting age
pi = 3.14159  # Approximate value of π

print(age)
print(pi)
````

# 3️⃣ Multi-Line Comments

Python **does not have a dedicated syntax** for multi-line comments.

The official and recommended way is to use multiple **`#`** symbols.

---

````python
# Step 1 : Read user input
# Step 2 : Process the data
# Step 3 : Display the result

print("Program Completed")
````

# ⚠️ Triple Quotes (`''' '''` or `""" """`)

Many beginners think triple quotes create multi-line comments.

**This is only partially correct.**

Technically,

Triple quotes create a **string literal**, **not a comment**.

If the string is **not assigned** to a variable or used as a docstring,

Python simply creates it and immediately discards it.

Therefore,

it behaves **like** a comment,

but it is **not an actual comment**.

---

````python
"""
This is NOT a real comment.

It is actually a string literal.

Python creates it and immediately discards it.
"""

print("Hello")
````

## How Python Actually Sees It

```
"""
Hello
World
"""
```

Python treats it like:

```
A string object

↓

Not stored anywhere

↓

Garbage Collected
```

So,

it behaves like a comment,

but technically,

it is **not** one.

---

# ✅ Official Recommendation

✔ Use **`#`** for comments.

✔ Use **Triple Quotes** only for **Docstrings**.

This is the recommendation followed by:

- Python Documentation
- PEP 8 Style Guide
- Professional Python Developers

---

# 4️⃣ Docstrings

A **Docstring** is a special string used to document

- Modules
- Functions
- Classes
- Methods

Unlike comments,

Python stores docstrings.

---


````python
def greet():
    """
    Print a welcome message.
    """
    print("Hello")
````

````python
print(greet.__doc__)
````

# 📊 Comments vs Docstrings

| Feature | Comments | Docstrings |
|----------|-----------|------------|
| Symbol | `#` | `""" """` |
| Executed | ❌ No | ✅ Stored |
| Purpose | Explain code | Documentation |
| Access with `__doc__` | ❌ No | ✅ Yes |
| Used by Documentation Tools | ❌ No | ✅ Yes |

---
