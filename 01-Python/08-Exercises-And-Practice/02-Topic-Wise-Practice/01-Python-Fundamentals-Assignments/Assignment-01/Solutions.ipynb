### Answer 6

```python
# Accept two inputs from the user
a = input("Enter first value (a): ")
b = input("Enter second value (b): ")

print(f"\nBefore swapping: a = {a}, b = {b}")

# Swap values using a temporary variable
temp = a
a = b
b = temp

print(f"After swapping: a = {a}, b = {b}")
```

### Answer 10

```python
# Take a decimal number as input
num = float(input("Enter a decimal number (e.g., 45.78 or -45.78): "))

# Extract the integer part using int() which cuts off towards zero
integer_part = int(num)

# Subtract the integer part from the original number to get the fraction
# round() handles floating-point precision quirks in binary arithmetic
fractional_part = round(num - integer_part, 10)

print(f"Original Number: {num}")
print(f"Integer Part: {integer_part}")
print(f"Fractional Part: {fractional_part}")
```

### Answer 12

```python
# Accept price and quantity inputs
price = float(input("Enter the price of one item: "))
quantity = int(input("Enter the quantity purchased: "))

# Calculate subtotal, 5% tax, and final total bill
subtotal = price * quantity
tax = subtotal * 0.05
total_bill = subtotal + tax

print(f"\n--- Shopping Receipt ---")
print(f"Subtotal: ${subtotal:.2f}")
print(f"Tax (5%): ${tax:.2f}")
print(f"Total Bill: ${total_bill:.2f}")
```

### Answer 15

```python
# Accept user distance and time parameters
distance = float(input("Enter distance traveled (in kilometers): "))
time = float(input("Enter travel time (in hours): "))

# Safely handle potential division by zero edge-case
if time > 0:
    speed = distance / time
    print(f"The average speed is {speed:.2f} km/h")
elif time == 0:
    print("Error: Travel time cannot be zero when moving a distance!")
else:
    print("Error: Time cannot be negative.")
```

### Answer 16

```python
# Get total seconds from user
total_seconds = int(input("Enter a total number of seconds: "))

# Extract whole hours
hours = total_seconds // 3600
# Find remaining seconds after extracting hours
remaining_seconds = total_seconds % 3600

# Extract minutes and final seconds from the remainder
minutes = remaining_seconds // 60
seconds = remaining_seconds % 60

print(f"{total_seconds} seconds is equal to: {hours}h {minutes}m {seconds}s")
```

### Answer 19

```python
# Accept numeric values
x = float(input("Enter first number (x): "))
y = float(input("Enter second number (y): "))

print(f"\nBefore swap: x = {x}, y = {y}")

# Python tuple unpacking option (cleanest & most idiomatic Python approach)
x, y = y, x

print(f"After swap: x = {x}, y = {y}")
```
