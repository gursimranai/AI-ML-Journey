### `Problem 1`: Write a program that will give you in hand monthly salary after deduction on CTC - HRA(10%), DA(5%), PF(3%) and taxes deduction as below:

> Salary(Lakhs) : Tax(%)

*   Below 5 : 0%
*   5-10 : 10%
*   10-20 : 20%
*   aboove 20 : 30%

````python
# Write code here
annual_ctc = int(input("Enter Your Annual CTC : "))

# Standard 18% deduction (HRA 10% + DA 5% + PF 3%) is factored into every multiplier
if annual_ctc < 500000:
    salary = annual_ctc * 0.82  # 0% Tax -> 100% - 18% = 82%
elif annual_ctc <= 1000000:
    salary = annual_ctc * 0.72  # 10% Tax -> 100% - 18% - 10% = 72%
elif annual_ctc <= 2000000:
    salary = annual_ctc * 0.62  # 20% Tax -> 100% - 18% - 20% = 62%
else:
    salary = annual_ctc * 0.52  # 30% Tax -> 100% - 18% - 30% = 52%

print(f"Your Salary In Hand Per Month: {round(salary/12, 2)}")
````

*Note: The Fibonacci Sequence is a series of numbers. The next number is found by adding up the two numbers before it. The first two numbers are 0 and 1. For example, 0, 1, 1, 2, 3, 5, 8, 13, 21. The next number in this series above is 13+21 = 34*

````python
# Write code here


first = 0
second = 1
count = 1

while (count <= 10):
    if (count == 1):
        print(first, end=",")
        count+=1
        continue
    elif (count == 2):
        print(second, end=",")
        count+=1
        continue
    else:
        new = first + second

        first = second
        second = new
        count+=1
        print(new,end=",")

    
````

### `Problem 6` - Find the factorial of a given number.

Write a program to use the loop to find the factorial of a given number.

The factorial (symbol: `!`) means to multiply all whole numbers from the chosen number down to 1.

For example: calculate the factorial of 5

```bash
5! = 5 × 4 × 3 × 2 × 1 = 120
```

Output:

```bash
120
```

````python
# Write code here
fact = 1

num = int(input("Enter The Number: "))

for x in range(num,1,-1):
    fact *= x

print(f"Factorial of {num} is {fact} !!!")
````

````python
# Step 1: Get user input
num = int(input("Enter a number: "))

factorial = 1

# Step 2: Handle negative numbers, 0, and positive numbers
if num < 0:
    print("Factorial does not exist for negative numbers.")
elif num == 0:
    print("The factorial of 0 is 1")
else:
    # Loop from 1 to num (inclusive)
    for i in range(1, num + 1):
        factorial = factorial * i
        
    print(f"The factorial of {num} is {factorial}")
````

````python
import math

# Step 1: Get user input
num = int(input("Enter a number: "))

# Step 2: Use the built-in math function
try:
    result = math.factorial(num)
    print(f"The factorial of {num} is {result}")
except ValueError:
    print("Factorial does not exist for negative numbers.")
````

````python
# Step 1: Define the recursive function
def find_factorial(n):
    # Base Case: If number is 0 or 1, return 1
    if n == 0 or n == 1:
        return 1
    # Recursive Case: n * factorial of (n - 1)
    else:
        return n * find_factorial(n - 1)

# Step 2: Take user input
num = int(input("Enter a number: "))

# Step 3: Check if input is valid and call the function
if num < 0:
    print("Factorial does not exist for negative numbers.")
else:
    result = find_factorial(num)
    print(f"The factorial of {num} is {result}")
````

### `Problem 7` - Reverse a given integer number.

Example:

`Input:`

```bash
76542
```

`Output:`

```bash
24567
```

````python
# Write code here
num = 143
rev = 0

while (num > 0):
    ones = num % 10
    rev = rev * 10 + ones

    num //= 10

print(rev)
````

````python
# Take the user input as an integer
num = int(input("Enter a number to reverse: "))

# Initialize the reversed number variable
reversed_num = 0

# Keep looping until the original number becomes 0
while num > 0:
    # 1. Get the last digit of the number (e.g., 123 % 10 = 3)
    last_digit = num % 10
    
    # 2. Add the digit to the reversed number (shift left by multiplying by 10)
    reversed_num = (reversed_num * 10) + last_digit
    
    # 3. Remove the last digit from the original number (e.g., 123 // 10 = 12)
    num = num // 10

# Print the final reversed number
print("Reversed Number:", reversed_num)
````

### `Problem 8`: Take a user input as integer N. Find out the sum from 1 to N. If any number if divisible by 5, then skip that number. And if the sum is greater than 300, don't need to calculate the sum further more. Print the final result. And don't use for loop to solve this problem.

**Example 1:**

`Input:`

```bash
30
```

`Output:`

```bash
276
```

````python
# Write code here

num = int(input("Enter the number: "))

current_num = 1
total_sum = 0

while current_num <= num:

    if current_num % 5 == 0:
        current_num+=1
        continue

    if current_num + total_sum > 300:
        break

    total_sum += current_num
    current_num +=1


print(f"Sum: {total_sum}")
````

````python
# Take the user input as an integer
N = int(input("Enter a number N: "))

# Initialize variables
total_sum = 0
current_number = 1

# Process numbers from 1 to N without using a for loop
while current_number <= N:
    # Rule 1: Skip if the number is divisible by 5
    if current_number % 5 == 0:
        current_number += 1
        continue
    
    # Rule 2: If adding the current number makes the sum exceed 300, stop entirely
    if total_sum + current_number > 300:
        break
        
    # Add the current number to the total sum
    total_sum += current_number
    current_number += 1

# Print the final result
print(total_sum)
````

###`Problem 11`: A robot moves in a plane starting from the original point (0,0). The robot can move toward UP, DOWN, LEFT and RIGHT with a given steps.
The trace of robot movement is shown as the following:
```
UP 5
DOWN 3
LEFT 3
RIGHT 2
!
```
> The numbers after the direction are steps.

> `!` means robot stop there.

**Please write a program to compute the distance from current position after a sequence of movement and original point.**

*If the distance is a float, then just print the nearest integer.*

Example:

`Input`:
```
UP 5
DOWN 3
LEFT 3
RIGHT 2
!
```
`Output`:
```
2
```

````python
# Write code here
from math import sqrt

pos = [0,0]
while True:
    s = input("Enter the robot path: ")

    if (s == '!'):
        break

    direction = s.split(' ')[0]
    step = int(s.split(' ')[1])

    if direction == 'UP':
        pos[1] = pos[1] + step
    elif direction == 'DOWN':
        pos[1] -= step
    elif direction == 'LEFT':
        pos[0] -= step
    elif direction == 'RIGHT':
        pos[0] += step
    else:
        pass

print(f"New Position: {pos}")
print(sqrt(pos[0]**2+pos[1]**2))


````

###`Problem 13`:Print all the Armstrong numbers in a given range.
Range will be provided by the user<br>
Armstrong number is a number that is equal to the sum of cubes of its digits. For example 0, 1, 153, 370, 371 and 407 are the Armstrong numbers.

````python
# Write code here

num = int(input("Enter the number: "))
len_num = len(str(num))
armstone_sum = 0
original_num = num


while (num > 0):
    digit = num % 10

    armstone_sum += digit ** len_num
    num //= 10

if (original_num == armstone_sum):
    print(f"{original_num} is an Armstrong number.")
else:
    print(f"{original_num} is NOT an Armstrong number.")
````

````python
# Take user input
num = int(input("Enter a number to check: "))

# 1. Store the original number to compare at the end
original_num = num

# 2. Find the total number of digits (order)
# Converting to string is the easiest way to count digits
num_digits = len(str(num))

# Initialize the variable to store the sum
armstrong_sum = 0

# 3. Extract digits and calculate power sums
while num > 0:
    # Get the last digit
    digit = num % 10
    
    # Raise the digit to the power of total digits and add to sum
    armstrong_sum += digit ** num_digits
    
    # Remove the last digit
    num = num // 10

# 4. Compare the result with the original number
if original_num == armstrong_sum:
    print(f"{original_num} is an Armstrong number.")
else:
    print(f"{original_num} is NOT an Armstrong number.")
````

###`Problem 14`:Calculate the angle between the hour hand and minute hand.

Note: There can be two angles between hands; we need to print a minimum of two. Also, we need to print the floor of the final result angle. For example, if the final angle is 10.61, we need to print 10.

Input:<br>
H = 9 , M = 0<br>
Output:<br>
90<br>
Explanation:<br>
The minimum angle between hour and minute
hand when the time is 9 is 90 degress.

````python
# Write code here

hour_hand = int(input("Enter the hour hand : "))
minute_hand = int(input("Enter the minute hand : "))

angle = 30 * hour_hand - 5.5 * minute_hand

degree = 360 - angle

print(f"Angle: {int(degree)}")
````

````python
import math

hour_hand = int(input("Enter the hour hand : "))
minute_hand = int(input("Enter the minute hand : "))

# 1. Fix the formula using subtraction and absolute value
angle = abs(30 * hour_hand - 5.5 * minute_hand)

# 2. Find the minimum (interior) angle dynamically
min_angle = min(angle, 360 - angle)

# 3. Use math.floor for the final result
degree = math.floor(min_angle)

print(f"Angle: {degree}")
````

###`Problem 15`:Given two rectangles, find if the given two rectangles overlap or not. A rectangle is denoted by providing the x and y coordinates of two points: the left top corner and the right bottom corner of the rectangle. Two rectangles sharing a side are considered overlapping. (L1 and R1 are the extreme points of the first rectangle and L2 and R2 are the extreme points of the second rectangle).

Note: It may be assumed that the rectangles are parallel to the coordinate axis.

<img src='https://www.geeksforgeeks.org/wp-content/uploads/rectanglesOverlap.png' width='300' height='200'>

````python
# Write code here
def check_rectangle_overlap(l1_x, l1_y, r1_x, r1_y, l2_x, l2_y, r2_x, r2_y):
    """
    This function checks if two rectangles overlap.
    
    Rectangle 1: Top-Left corner is (l1_x, l1_y) and Bottom-Right corner is (r1_x, r1_y)
    Rectangle 2: Top-Left corner is (l2_x, l2_y) and Bottom-Right corner is (r2_x, r2_y)
    
    Note: This assumes standard screen coordinates where:
    - X gets larger as you move RIGHT
    - Y gets larger as you move DOWN
    """
    
    # CONDITION 1: Check if one rectangle is completely to the LEFT of the other
    # If the right edge of Rectangle 1 is to the left of the left edge of Rectangle 2...
    # OR if the right edge of Rectangle 2 is to the left of the left edge of Rectangle 1...
    # Then they cannot overlap!
    if r1_x < l2_x or r2_x < l1_x:
        return False

    # CONDITION 2: Check if one rectangle is completely ABOVE the other
    # If the bottom edge of Rectangle 1 is higher up (smaller Y) than the top edge of Rectangle 2...
    # OR if the bottom edge of Rectangle 2 is higher up (smaller Y) than the top edge of Rectangle 1...
    # Then they cannot overlap!
    if r1_y < l2_y or r2_y < l1_y:
        return False

    # If neither of the above conditions is true, it means they MUST be overlapping!
    return True


# =====================================================================
# EXAMPLE USAGE: Let's test the function with actual numbers
# =====================================================================

# Define coordinates for Rectangle 1 (Green rectangle in your diagram)
# Top-Left corner (l1) and Bottom-Right corner (r1)
rect1_left_x = 0
rect1_top_y = 0
rect1_right_x = 10
rect1_bottom_y = 8

# Define coordinates for Rectangle 2 (Black rectangle in your diagram)
# Top-Left corner (l2) and Bottom-Right corner (r2)
rect2_left_x = 5
rect2_top_y = -4
rect2_right_x = 15
rect2_bottom_y = 4

# Call the function and store the result (True or False)
is_overlapping = check_rectangle_overlap(
    rect1_left_x, rect1_top_y, rect1_right_x, rect1_bottom_y,
    rect2_left_x, rect2_top_y, rect2_right_x, rect2_bottom_y
)

# Print the final result out to the screen
if is_overlapping:
    print("The rectangles overlap!")
else:
    print("The rectangles do not overlap.")
````
